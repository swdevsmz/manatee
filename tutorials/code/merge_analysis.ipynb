# ハッシュ化データの突合サンプル
purchases.csv（購買データ）とexposures.csv（広告接触データ）をハッシュ化IDで突合します。

```python
import pandas as pd

# データ読み込み
df_purchases = pd.read_csv('purchases.csv')
df_exposures = pd.read_csv('exposures.csv')

# ハッシュ化IDで突合（内部結合）
merged = pd.merge(df_purchases, df_exposures, on=['user_id','email','phone'], how='inner', suffixes=('_purchase', '_exposure'))
print('突合結果:')
print(merged)
```

## 分析例
- 広告接触後に購買したユーザーの購買金額や日付を分析できます。
- さらに集計や可視化も可能です。
